# 🌱 Experiment 3: Exploring Distributions

An average summarizes a group with one number, but it doesn’t show how the individual measurements vary.

Are most plants close to the average height? Or does the average fall between many short plants and many tall ones?

The pattern of values—and how frequently they occur—is called a **distribution**.

## 📊 Meet the Histogram

A **histogram** groups numerical measurements into intervals called **bins**.

- The horizontal axis shows the measurement values.
- The vertical axis shows how many observations fall within each bin.
- Taller bars indicate intervals containing more observations.

Let’s examine the heights of the bean plants we measured, combining all three density treatments for this first look.

## 💻 Step 1: Graph the Measurements

⬇️ Run the code below before continuing.

```python
# Select height measurements for beans
bean_heights = df[
    (df["Species"] == "Bean") &
    (df["Metric"] == "Height")
]

# Create a histogram
plt.figure(figsize=(8, 5))

plt.hist(
    bean_heights["Value"],
    bins=10,
    color="seagreen",
    edgecolor="white"
)

plt.xlabel("Plant height (cm)")
plt.ylabel("Number of plants")
plt.title("Distribution of Bean Plant Heights")

plt.show()
```

## 👀 Read the Graph

Each plant contributes one height measurement to this histogram. Plants whose heights fall within the same bin are counted together.

> 💡 **Computational Note**
>
> `bins=10` divides the range of observed heights into ten equal-width intervals. It does **not** put ten plants in each bar.

## 🌱 Does Density Change the Picture?

Our first histogram combined bean heights from all three density treatments.

Could combining those measurements hide differences among treatments?

Let’s separate the data into **Low**, **Med**, and **High** density.

## 🤔 Predict

How do you expect plant heights to differ among density treatments?

## 💻 Step 2: Compare Density Treatments

The graphs below use the same height intervals and axis scales so we can compare them consistently.

⬇️ Run the code below before continuing.

```python
import numpy as np

# Use the same ten bins for all treatments
bin_edges = np.linspace(
    bean_heights["Value"].min(),
    bean_heights["Value"].max(),
    11
)

# Create three graphs with matching axes
fig, axes = plt.subplots(
    1, 3,
    figsize=(12, 4),
    sharex=True,
    sharey=True
)

for ax, density in zip(axes, ["Low", "Med", "High"]):
    heights = bean_heights.loc[
        bean_heights["Density"] == density,
        "Value"
    ]

    ax.hist(
        heights,
        bins=bin_edges,
        color="seagreen",
        edgecolor="white"
    )

    ax.set_title(f"{density} density (n = {len(heights)})")
    ax.set_xlabel("Plant height (cm)")

axes[0].set_ylabel("Number of plants")
fig.suptitle("Bean Plant Heights by Density")

plt.tight_layout()
plt.show()
```

## 👀 Compare the Distributions

The **n** above each graph tells you how many plant height measurements it contains.

These sample sizes differ. A treatment with more measured plants can have taller bars simply because it contains more observations. Consider where the measurements fall and how widely they are spread, as well as bar height.

These graphs describe individual plant measurements. When we test for a density effect, we’ll account for plants sharing a pot by calculating **one mean per pot**.

## 🪴 Step 3: What Counts as a Replicate?

Our histograms showed individual plant heights. But plants growing together in a pot share soil, water, space, and neighbors.

Their growing conditions are connected.

For our density experiment, the **pot is the experimental unit**: the unit that received a density treatment. Plants within that pot are measurements within the same replicate.

Ten plants in one pot therefore do not provide the same independent replication as ten plants in ten separate pots.

To compare density treatments, we’ll calculate **one average height per pot**. Each pot will then contribute one value to our analysis.

If a pot has fewer recorded measurements, we’ll calculate its mean from the measurements available.

## 💻 Calculate the Pot Means

⬇️ Run the code below before continuing.

```python
# Calculate average bean height within each pot
bean_pot_heights = (
    bean_heights
    .groupby(["Density", "Pot ID"], as_index=False)
    .agg(
        Mean_height_cm=("Value", "mean"),
        Plants_measured=("Value", "count")
    )
)

# Display the first few pot summaries
bean_pot_heights.head().round(2)
```

## 👀 Take a Look

Each row in this new table represents **one pot**.

- **Mean_height_cm** is the average height of the measured plants in that pot.
- **Plants_measured** tells us how many height measurements contributed to that average.

For a low-density pot containing one plant, the pot mean is simply that plant’s height.

> 💡 **Computational Note**
>
> `groupby()` gathers rows with matching density and pot labels. The code then calculates the mean and counts the recorded measurements within each group.

Let’s check how many pots contribute to each density treatment.

```python
# Count the pots contributing to each density treatment
for density in ["Low", "Med", "High"]:
    number_of_pots = (
        bean_pot_heights["Density"] == density
    ).sum()

    print(f"{density} density: {number_of_pots} pots")
```

## 🤔 Think

- Why does averaging within pots help us avoid counting plants that share a pot as independent replicates?
- Does a pot containing ten measured plants contribute more independent replicates than a pot containing five?
- What information about individual plants is no longer visible when we summarize each pot with a mean?

We now have one height value per pot. Next, we’ll graph these values to compare the density treatments.

## 📊 Step 4: Compare the Pot Means

Each point in the next graph represents **one pot’s average bean height**.

This graph lets us compare:

- where the pot means fall in each treatment;
- how much they vary within treatments;
- how much the treatments overlap.

The points are spread slightly from side to side so they are easier to see. Their horizontal positions within a treatment have no biological meaning.

⬇️ Run the code below before continuing.

```python
# Plot one mean height per pot
fig, ax = plt.subplots(figsize=(8, 5))

densities = ["Low", "Med", "High"]

for position, density in enumerate(densities):
    pot_means = bean_pot_heights.loc[
        bean_pot_heights["Density"] == density,
        "Mean_height_cm"
    ].to_numpy()

    # Spread points horizontally to reduce overlap
    offsets = (
        np.linspace(-0.12, 0.12, len(pot_means))
        if len(pot_means) > 1
        else np.zeros(len(pot_means))
    )

    ax.scatter(
        position + offsets,
        pot_means,
        color="seagreen",
        s=65,
        alpha=0.8
    )

ax.set_xticks(range(3))
ax.set_xticklabels(densities)
ax.set_xlabel("Density treatment")
ax.set_ylabel("Mean plant height per pot (cm)")
ax.set_title("Bean Height: One Point per Pot")

plt.tight_layout()
plt.show()
```

## ⚖️ From Patterns to Evidence

Differences among treatment averages can arise even when a treatment has no effect. Individual pots vary for many reasons.

Our next question is:

**Are the differences among density treatments large relative to the variation among pots within treatments?**

A statistical method called **analysis of variance**, or **ANOVA**, helps us evaluate that question.

Before using it, we’ll identify why it fits our question and consider its assumptions.
